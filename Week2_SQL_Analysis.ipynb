```python
import pandas as pd
import sqlite3

# Load the sales dataset
df = pd.read_csv("SQL_Sales_Dataset.csv")

# Create a SQLite database connection
conn = sqlite3.connect("sales_database.db")

# Store the dataset as a SQL table
df.to_sql("sales", conn, if_exists="replace", index=False)

print("Dataset successfully loaded into SQLite database!")
```

```python
query = """
SELECT *
FROM sales
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT *
FROM sales
WHERE region = 'West';
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    category,
    SUM(total_price) AS total_revenue
FROM sales
GROUP BY category
ORDER BY total_revenue DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    SUM(total_price) AS total_sales,
    AVG(total_price) AS average_sales,
    COUNT(*) AS total_records
FROM sales;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT *
FROM sales
ORDER BY category ASC, total_price DESC;
"""

result = pd.read_sql_query(query, conn)
display(result.head(10))
```

```python
query = """
SELECT
    customer_name,
    SUM(total_price) AS total_revenue
FROM sales
GROUP BY customer_name
ORDER BY total_revenue DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    customer_name,
    SUM(total_price) AS total_revenue
FROM sales
GROUP BY customer_name
ORDER BY total_revenue DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    AVG(order_total) AS average_order_value
FROM (
    SELECT
        order_id,
        SUM(total_price) AS order_total
    FROM sales
    GROUP BY order_id
);
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    order_id,
    customer_name,
    total_price,
    CASE
        WHEN total_price >= 1000 THEN 'High'
        WHEN total_price >= 500 THEN 'Medium'
        ELSE 'Low'
    END AS sales_category
FROM sales
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
customers = df[['customer_name']].drop_duplicates()

customers.to_sql('customers', conn, if_exists='replace', index=False)

print("Customers table created successfully!")
```

```python
query = """
SELECT
    sales.order_id,
    sales.customer_name,
    sales.total_price
FROM sales
INNER JOIN customers
ON sales.customer_name = customers.customer_name
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
display(result)
```
